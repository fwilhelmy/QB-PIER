```python
import pandas as pd

# Ask user for a CSV file path
file_path = input("Please enter the CSV file path: ")

# Load the CSV file into a DataFrame
try:
    df = pd.read_csv(file_path)
    print("DataFrame loaded successfully! The shape is:", df.shape)
except Exception as e:
    print("Error loading DataFrame:", e)
```

```python
second_file_path = input("Please enter the second CSV file path: ")

try:
    second_df = pd.read_csv(second_file_path)
    print("Second DataFrame loaded successfully! The shape is:", second_df.shape)
except Exception as e:
    print("Error loading second DataFrame:", e)
```

```python
# Filter rows where seed is not 0 (i.e., keeping only seed 42)
df_filtered = df[df['seed'] == 0]
second_df_filtered = second_df[second_df['seed'] == 0]

print("Filtered df shape:", df_filtered.shape)
print("Filtered second_df shape:", second_df_filtered.shape)
```

```python
merged_df = pd.concat([df_filtered, second_df_filtered]).drop_duplicates(
    subset=["seed", "source_model", "target_model", "epsilon"], keep="last"
)
print("Merged DataFrame shape:", merged_df.shape)
```

```python
print("Unique source models in df:", df['source_model'].unique())
print("=============================================")
print("Unique target models in df_filtered:", df_filtered['target_model'].unique())
print("=============================================")
print("Unique target models in second_df:", second_df['target_model'].unique())
print("=============================================")
print("Unique target models in second_df_filtered:", second_df_filtered['target_model'].unique())
print("=============================================")
print("Unique target models in merged_df:", merged_df['target_model'].unique())
```

```python
def rename_ampvqc(model_str):
    if "AmpVQC/model.n_qdepth=50" in model_str:
        return "AmpVQC50"
    elif "AmpVQC/model.n_qdepth=100" in model_str:
        return "AmpVQC100"
    elif "AmpVQC" in model_str:
        return "AmpVQC200"
    return model_str

merged_df['source_model'] = merged_df['source_model'].apply(rename_ampvqc)
merged_df['target_model'] = merged_df['target_model'].apply(rename_ampvqc)
```

```python
print("=============================================")
print("Unique target models in merged_df:", merged_df['target_model'].unique())
```

```python
merged_df.to_csv("merged_data.csv", index=False)
print("merged_data.csv saved successfully!")
```
